```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk


# Beispieltext
text = "Der Hund bellt laut. Die Katze miaut leise.".lower()
tokens = word_tokenize(text)

# Modell trainieren (einfaches Beispiel)
model = Word2Vec([tokens], vector_size=50, window=2, min_count=1, sg=1)

# Ähnliche Wörter
print(model.wv.most_similar("hund"))

# Vektor eines Wortes
print(model.wv["hund"])
```

GloVe

GloVe ist nur auf Englisch trainiert. 

```python
import gensim.downloader as api

# Lade GloVe-Modell (100-dim)
glove = api.load("glove-wiki-gigaword-100")

# Beispiel: Ähnlichste Wörter zu "dog"
print(glove.most_similar("dog"))

# Vektor für "dog"
print(glove["dog"])
```

BERT

kontextsensitiv

```python
import numpy as np
np.object = object   
np.bool = bool   
np.int = int  
np.float = float    
from transformers import BertTokenizer, BertModel
import torch

# Lade deutsches BERT-Modell
tokenizer = BertTokenizer.from_pretrained("bert-base-german-cased")
model = BertModel.from_pretrained("bert-base-german-cased")

# Tokenisieren & Tensor erstellen
text = "Der Hund bellt laut."
inputs = tokenizer(text, return_tensors="pt")

# Berechne Embeddings
with torch.no_grad():
    outputs = model(**inputs)

# Vektor für das Wort "Hund" extrahieren
token_index = inputs["input_ids"][0].tolist().index(tokenizer.convert_tokens_to_ids("Hund"))
vector = outputs.last_hidden_state[0][token_index]

print("BERT-Vektor für 'Hund':", vector)
```
